```python
# ===== 案例1: 共享内存 =====
t1 = torch.rand(2,3)
n1 = t1.numpy()          # 关键：numpy() 返回共享同一块内存的视图
t1[0,0] = 100            # 改 t1    
n1[1,1] = 999            # 改 n1
print(t1)               # t1[1,1] 也变成 999
print(n1)               # n1[0,0] 也变成 100
```

```python
import torch

t1 = torch.rand(2,3)
print(t1)

print(type(t1))


# 不共享内存
n1  = t1.numpy().copy()
print(n1)
print(type(n1))


print('--------------- 修改 t1的数据  会导致 n1 也修改 不共享内存----------------------')

# 修改 t1的数据  不会导致 n1 也修改
t1[0,0] = 100

print(t1)
print(n1)
```
